```python

"""
Student Management System
Features:
1. Add student
2. Update student
3. Delete student
4. List all students
5. Unique student ID validation
6. Save and load records using JSON
7. Formatted console output
"""

import json
import os


DATA_FILE = "students.json"


class Student:
    """Represents a single student."""

    def __init__(self, student_id, name, grade):
        self.student_id = student_id
        self.name = name
        self.grade = grade

    def to_dict(self):
        """Convert student object into a dictionary."""
        return {
            "student_id": self.student_id,
            "name": self.name,
            "grade": self.grade
        }


class StudentManager:
    """Manages student records and file operations."""

    def __init__(self, filename=DATA_FILE):
        self.filename = filename
        self.students = {}
        self.load_students()

    def load_students(self):
        """Load student records from JSON file."""
        if os.path.exists(self.filename):
            try:
                with open(self.filename, "r", encoding="utf-8") as file:
                    data = json.load(file)

                for record in data:
                    student = Student(
                        record["student_id"],
                        record["name"],
                        record["grade"]
                    )
                    self.students[student.student_id] = student

            except (json.JSONDecodeError, KeyError, TypeError):
                print("Warning: Could not read student data.")

    def save_students(self):
        """Save all student records to JSON file."""
        with open(self.filename, "w", encoding="utf-8") as file:
            data = [
                student.to_dict()
                for student in self.students.values()
            ]
            json.dump(data, file, indent=4)

    def add_student(self, student_id, name, grade):
        """Add a student if the ID is unique."""
        student_id = student_id.strip()
        name = name.strip()
        grade = grade.strip()

        if not student_id or not name or not grade:
            print("Error: All fields are required.")
            return

        if student_id in self.students:
            print("Error: Student ID already exists.")
            return

        student = Student(student_id, name, grade)
        self.students[student_id] = student
        self.save_students()
        print("Student added successfully!")

    def update_student(self, student_id):
        """Update an existing student's information."""
        student_id = student_id.strip()

        if student_id not in self.students:
            print("Error: Student not found.")
            return

        student = self.students[student_id]

        name = input(f"Enter new name [{student.name}]: ").strip()
        grade = input(f"Enter new grade [{student.grade}]: ").strip()

        if name:
            student.name = name
        if grade:
            student.grade = grade

        self.save_students()
        print("Student updated successfully!")

    def delete_student(self, student_id):
        """Delete a student using their ID."""
        student_id = student_id.strip()

        if student_id not in self.students:
            print("Error: Student not found.")
            return

        del self.students[student_id]
        self.save_students()
        print("Student deleted successfully!")

    def list_students(self):
        """Display all student records in a formatted table."""
        if not self.students:
            print("\nNo student records found.")
            return

        print("\n" + "=" * 58)
        print(f"{'STUDENT ID':<15}{'NAME':<25}{'GRADE':<15}")
        print("-" * 58)

        for student in self.students.values():
            print(
                f"{student.student_id:<15}"
                f"{student.name:<25}"
                f"{student.grade:<15}"
            )

        print("=" * 58)


def main():
    manager = StudentManager()

    while True:
        print("\n===== STUDENT MANAGEMENT SYSTEM =====")
        print("1. Add Student")
        print("2. Update Student")
        print("3. Delete Student")
        print("4. List Students")
        print("5. Exit")

        choice = input("Enter your choice (1-5): ").strip()

        if choice == "1":
            student_id = input("Enter student ID: ")
            name = input("Enter student name: ")
            grade = input("Enter student grade: ")

            manager.add_student(student_id, name, grade)

        elif choice == "2":
            student_id = input("Enter student ID to update: ")
            manager.update_student(student_id)

        elif choice == "3":
            student_id = input("Enter student ID to delete: ")
            manager.delete_student(student_id)

        elif choice == "4":
            manager.list_students()

        elif choice == "5":
            print("Thank you for using Student Management System!")
            break

        else:
            print("Invalid choice. Please enter a number from 1 to 5.")


if __name__ == "__main__":
    main()
```
